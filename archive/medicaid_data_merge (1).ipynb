```python
import pandas as pd

# Load the two xlsm datasets
df1 = pd.read_excel('drug_utilization_2024.xlsm')
df2 = pd.read_excel('drug_utilization_product.xlsm')

# Perform a left join on 'NDC Code'
merged_df = pd.merge(df1, df2, on='NDC Code', how='left')

# Save the merged dataset to CSV
merged_df.to_csv('merged_output.csv', index=False)

print("✅ Merge complete! File saved as 'merged_output.csv'")
```
